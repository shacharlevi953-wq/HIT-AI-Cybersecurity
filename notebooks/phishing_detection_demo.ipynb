# AI-Based Phishing Website Detection
הדגמה קצרה של טעינת תוצאות הניסוי והצגת המדדים המרכזיים. להרצה מלאה השתמשו בקובץ `src/phishing_detection.py`.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import Image, display

repo = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
results = repo / 'results'
```

```python
metrics = json.loads((results / 'metrics.json').read_text(encoding='utf-8'))
pd.Series({
    'Accuracy': metrics['accuracy'],
    'Phishing precision': metrics['phishing_precision'],
    'Phishing recall': metrics['phishing_recall'],
    'Phishing F1': metrics['phishing_f1'],
    'ROC-AUC': metrics['roc_auc'],
}).map(lambda x: f'{x:.2%}')
```

```python
display(Image(filename=str(results / 'confusion_matrix.png')))
display(Image(filename=str(results / 'roc_curve.png')))
display(Image(filename=str(results / 'feature_importance.png')))
```
